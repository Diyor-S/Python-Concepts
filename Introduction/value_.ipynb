In previous lesson, we have covered Object's **type**. We concluded that we should think of `type` as `operations` that we can do with the object.

In this lesson, we are talking about the **third** component of the Object - its **value**.

This one is interesting as well.
Consider this example:

```python
number1 = 10 # While this is clear, the number has value of 10. This is crystal clear.
print(number1)
```

```python
string1 = "Some text" # Here, we also have it crystal clear, the value of the object of type 'str' is "Some text"
print(string1)
```

```python
some_list = [1, 2, 3] # Here it is also clear
some_list
```

```python
some_list = ["c", "t", "a"] # Again crystal clear the value of this list.
some_list
```

What conclusion we should make?  
While there are objects that we are able to recognize easily such as: numbers, strings, lists, tuples, and so on,  
there are also other `types` of Objects like:  
functions, classes, methods, whose `value` is not that clear, because we are not that `familiar` with them in our everyday lives   
as opposed to numbers, and strings.

For now, I just wanted to show the fact, that it is `Developers` of `Python`, 
who `decided` to `define` what would be the object's `value`!

They used `Logic` to define what would be appropriate to use as the `value` of the Object!
It is logically correct to represent the `number` itself for Objects of **type** `int`.  
It is logically correct to represent the `text` itself for Objects of **type** `str`.

Every other programming language also `represents` the `value` of the Object in its own way.

Remember that I said that `type` is all about `operations` that we can do with that `Object`,   
what we are actually performing our operations on is `value`,    
we change the value of the Object whenever we talk about `type` -> `operations`.   

```python
string1 = "STRING" # value of the Object is all uppercase characters.

string1.lower() # we use `operation - lower()` on our Object of type 'str'. 
# And we `performed` that operation on `value` and got modified result 'string'!

# So why we can perform lower() operation on an object of type 'str'?
# And if we do this:
```

```python
num1 = 10

num1.lower() # We get an error? 'AttributeError' this is extremely important Exception to engrave in mind.
# Objects of type 'int' do NOT support operation 'lower()'.
```

```python
list1= [1, 2, 3]
list1.lower() # The same error! AttributeError! Objects of type 'list' do NOT support operation 'lower()'.
```

What can we understand from previous examples?   
We should conclude that it is the `type` of the Object that defines what operations we can perform on this Object's `value`!

So there is a strong relationship between the `type` and `value` of the Object!
We must not forget about `identity (id)` of the Object as well.

# What is an **Object**?

When I first came into programming, I did not even think about this **fundamental concept**.
I heard from my professor that `Everything in Python is an Object`. I said like, okay, noted, and then forgot.

When we talk about programming - on the base level, we are instructing the **machine** to do **something**.
We humans understand that there are **Objects** like: Table, Chair, Window, Door, Car, etc.

But what the machine that operates only on 0s and 1s understand? 
How can we explain to it that there is a Table object, Chair object, Window object?
We can't do that.

But you know how developers did that?
They just said: we can **represent** anything as an **Object**, and it would have **3** things in common: `id, type, value`.

Look, we have our **mind** to keep the **context**. 
The fact that we **know** about **some concept** means that there is a need to `store` it `somewhere - mind`.

That is why **computers** also need `somewhere (RAM)` to store variables that we defined.

Anyways, those 3 things: id, type, and value are `important components` of the Object to keep in mind.

```python
# How does an object look like in Python?
# Before discuss that, look at this "Exception - NameError".
a
print(id(a))
```

Why Python raised `Exception` `NameError` straight away? We haven't even done anything yet, and Python already restricts us...

Turns out, that there is a concept of `Identifier`. And this concept is `NOT` unique to Python. Every other programming language has it.
This concept of `Identifier` is what we often refer as `Variable` but they are not the same thing. 

Anyways, the reason Python three that `NameError` is that, `a` above is `NOT` an `Object`. It is an `Identifier`.

What is an `Identifier`? It is just a `name` that we give to so called **variables** to `ACCESS` the **Object's** `value`.

Consider Examples below:

```python
'a' # This is now an 'object'. Its 'type' is 'str'.
```

```python
print('a') # value -> a
print(id('a')) # id -> address in the memory
print(type('a')) # type -> str 
```

```python
1 # The same goes for the number, its 'type' is 'int'.
```

```python
print(1)
print(id(1))
print(type(1))
```

```python
a = "This is string"
a # a - is an identifier, while the `Object`` is on the right part of assignment operator.
# That is why we say: `name` "a" refers to "object" with type `str` and `value` "This is string"
```

```python
num = 5
num # The same goes to identifier "num",  name "num" refers to object with type 'int' and value '5'.
```

```python
$s = 5 # Notice that SyntaxError Exception got raised when we put $ sign in front of 's' character.

# It means that there are "RULES" that Python Interpreter uses to actually distinguish
# "VALID" Identifier and an "INVALID" one which raises SyntaxError.
```

You can learn more about Identifiers and their rules here:  
https://docs.python.org/3/reference/lexical_analysis.html#names-identifiers-and-keywords:~:text=2%2E3%2E%20Names

I hope that it is now clear that **Object** consists of **three** things: `id, type, value`.   
The `type` is the most interesting one in my opinion.  
And **Identifier** is just a name, we give to an **Object**, to `access` its **value**.

What is an `Exception`?

Before answering that,

What is a difference between an `Error` and an `Exception`?

Anyway,     
Error - this is something that has `resulted` in some kind of `UNEXPECTED` behaviour of the program.  
Exception - is something that has `resulted` in some behaviour of the program that **WE (developer)** actually `EXPECTED`.  

This makes in fact a huge difference.

Consider this example:

```python
# This should raise ZeroDivisionError, 
# and yeah the "Error" at the end should not make you think that this is unexpected.
# Precisely because we can not divide the numbers by zero in (basic math), 
# Python devs already handled this Exception for us. 
num1 = 1 / 0
```

```python
num1 = 5
# Look at another Exception, AttributeError, again Error should not make us think that it was unexpected.
num1.lower()
```

```python
num1 = 1
string1 = "s"
# Let's have a look at one more Exception, TypeError.
num1 + string1
```

And yes, there are tons of Exceptions in Python, you can have a look at them in python documentation.

But the core ones are:

`TypeError`  
`ValueError`  
`AttributeError`  

These above, are the most important ones in my view that are provided to us by Python Core Devs.

Does `TypeError` sounds familiar? It is deeply related to Object's `type`.

I think I would phrase Exceptions as: 
"You did something that I do not `accept`, please `fix that`, and give me `correct data`.

Here `You` refers to the **users** of the programming language in our case **we** use **Python**,  

`I` refers to the **programming language** itself,   

While `fix that` refers to the fact that Python **raises** an `Exception`:
this is `Python's` way of saying that **we (users)** need to **do** something **differently**.


`Correct data` refers to `ALLOWED` **actions/operations** that we **can** do with the Object.

Today, we are gonna talk about `TypeError` in a bit more detail.

What is `TypeError` **Exception**?

`TypeError` is an Exception raised when some invalid input has been received regarding the Object's `type`.   

Here, to understand this **Exception**,   
we recall that Object's **type** defined the `operations` that we could perform on an Object's **value**.

Let's consider this example:

```python
# Our basic example of trying to add number with a string
num1 = 5
string1 = "s"

num1 + string1 # So, there is a useful message within the `TypeError` Exception of what exactly we got wrong.

# It says unsupported operand '+' between Object types 'int' and 'str'
```

```python
result = 5 + 5
result # No issue with numbers, we can add them together
```

```python
# It works even with float numbers
result = 5 + 4.0
result # But the result is `float`.
```

```python
"string1" - "str1" # Again, the same pattern repeats. Operation '-' is not supported for Objects of type 'str'.
```

```python
[10, 20, 30] - [1, 2, 3] # The same is applied for Object's of 'type' list. Tho we have numbers, 
# but they are wrapped into a lists. And lists do not support operation '-' between each other. 
```

```python
[10, 20, 30] * [1, 2, 3] # Notice the message, the message says it all. It did not say, that it does not support *.
# It asks specifically for the number to be multiplied!
```

```python
# So if we use * <number>, the list should accept this.
[10, 20, 30] * 2 # So, look at the output, the language itself guides us toward
# what things we can perform with the value of a particular type.
```

```python
# The same goes to the 'dict'
{1: "one", 2: "two"} - {1: "one"}
```

```python
{1: "one", 2: "two"} * {1: "one"}
```

So what we should understand from those experiments made above?

First things first, we need to use Exceptions, we need to make as many `mistakes` as possible. 

By this, I meant try out the things and getting more of those Exceptions.   
That way we actually improve our skill of recognizing the same pattern again and again, and we learn to avoid those problems,   
and most importantly we `spot` those possible `Exceptions` just by looking at the code itself.

How do you think `Professional Developers` can spot the issues in large codebases?

They have learned the patterns that lead to the bugs and issues later on, and because they have already made a lot mistakes  
they gain the ability to spot those problems even before the code gets executed and the issue is discovered during runtime.  

This specific `Experiment` with `TypeError` should teach us things like:  
1. Knowing that there is a possible Exception like `TypeError`  
2. Even if we made such mistake and `TypeError` Exception has been raised, we know `WHY` it was raised,  
meaning we know the `scope` of the problem, we can quickly find and fix that issue.  
3. It teaches us the pattern that there are dozens of Exceptions that convey different meanings (different scope).  

Today's `scope` - `TypeError` Exception means that there is some operation that `we tried` to `perform` on the Object's value,  
whose type did not `define` such operation `possible` to be used.

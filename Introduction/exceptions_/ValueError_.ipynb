that means something invalid has been attempted to be performed on Object's value.   
Ultimately, violation occured with the Object's `type`.  

Today, we are talking about closely related and fairly popular another **Exception** `ValueError`.

What is **ValueError**?

So far we have understood that `TypeError` occurs when the scope of the issue is within the Object's **type**.

And `ValueError` - is an **Exception** that narrows down the scope of our issue to Object's **value**. 

What does that mean, you may ask.

While Object's **type** defines the **possible** set of `operations` that can be **performed** over the Object's **value**,
Object's **value** defines the possible set of `VALID` values that are **ACCEPTABLE**.

Consider this example:

```python
num1 = int("1") # We are passing an Object of type 'str' to the 
# int let's call it factory that gets new Objects of type 'int' for now.
num1 # It works fine
```

```python
num2 = int("s") # However, if we do not pass the number in the string format, but a char
# We instantly get ValueError
# Why do we get ValueError? The reason is that,
# while "1" - number one wrapped into "" quotes which turns it into Object of type 'str' CAN be CONVERTED to int Object,
# The "s" - is not a number at all, it is a character. That is a fundamental difference. 
```

```python
num3 = int("2.0") # And tho float numbers CAN be represented as integers, Python still raises ValueError because 
# when we pass str , Python expects only integers - whole numbers to convert them into numbers (to type 'int') from type 'str'
```

So Object's of **type** `int` CAN accept Object's of type 'str', but they MUST BE whole integers.

This is a very good example of the fact that Python says:
Yes you can use **objects of type `str`**, but I define the range of `acceptible` values.

Let's consider another example:

```python
list1 = [1, 2, 3] # We should know that we have `indices` that we can use to access the elements of the list
print(list1[0]) # First element
print(list1[1])
print(list1[2])
```

```python
# But we also can find the some Value's index
print(list1.index(1)) # First element's index = 0
print(list1.index(3)) # Last element's index = 2
```

```python
# Notice what we get if we try to access the index of value that is not present in the list:
list1.index(5) # There is no element 5 within our list1
```

```python
# Another example:

float1 = float(2)
float1
```

```python
float2 = float("2")
float2
```

```python
float3 = float("5.0")
float3 # Notice the difference, our 'int' did not accept float within the str, but type 'float' finds it acceptable!
```

```python
float4 = float("s") # If we again try to pass as 'str' - character, we are gonna get the same ValueError
float4 # Notice what the Error message says: could not CONVERT.
```

What conclusion can we make from the little experiments we conducted?

ValueError - is an Exception that is raised whenever the `UNACCEPTABLE` value was **attempted** to be `assigned`.

If the `TypeError` is raised for having issues with the `operations` on the Object's **value**, 
`ValueError` is raised whenever we pass the `WRONG/UNACCEPTABLE` value to perform that **operation** on!

Today, we are gonna talk about `Exceptions`. This is an **EXTREMELY** important concept.   
This is our guide that guides us through the language, in fact, not only in Python, but in any other programming language.

What is an `Exception`?

Before answering that,   
one person who had worked at `Yandex` as a Python Developer when conducting a mock interview with some candidate,     
he asked these questions:     
What is a difference between an `Error` and an `Exception`?     
Are they the `same` thing or **factually different**?    
 
And this was a thought-provoking question for me,     
because I realized that I missed the different meaning that those two things tried to convey.  

Maybe that was because I am not native English speaker...  

Anyway,     
Error - this is something that has `resulted` in some kind of `UNEXPECTED` behaviour of the program.  
Exception - is something that has `resulted` in some behaviour of the program that **WE (developer)** actually `EXPECTED`.  

This makes in fact a huge difference.

Consider this example:

```python
# This should raise ZeroDivisionError, 
# and yeah the "Error" at the end should not make you think that this is unexpected.
# Precisely because we can not divide the numbers by zero in (basic math), 
# Python devs already handled this Exception for us. 
num1 = 1 / 0
```

```python
num1 = 5
# Look at another Exception, AttributeError, again Error should not make us think that it was unexpected.
num1.lower()
```

```python
num1 = 1
string1 = "s"
# Let's have a look at one more Exception, TypeError.
num1 + string1
```

And yes, there are tons of Exceptions in Python, you can have a look at them in python documentation.

But the core ones are:

`TypeError`  
`ValueError`  
`AttributeError`  

These above, are the most important ones in my view that are provided to us by Python Core Devs.

Does `TypeError` sounds familiar? It is deeply related to Object's `type`.

I think I would phrase Exceptions as: 
"You did something that I do not `accept`, please `fix that`, and give me `correct data`.

Here `You` refers to the **users** of the programming language in our case **we** use **Python**,  

`I` refers to the **programming language** itself,   

While `fix that` refers to the fact that Python **raises** an `Exception`:
this is `Python's` way of saying that **we (users)** need to **do** something **differently**.


`Correct data` refers to `ALLOWED` **actions/operations** that we **can** do with the Object.

Before we talk about the **second** constituent of the Object - **type**,   
I would like to mention one thing that I forgot to mention in previous lesson.

This is about the `FUNDAMENTAL` concept of `Dynamic Languages`. Yes, Python is a `dynamically typed` language. Just like Javascript.
Unlike Java, C, C++, Python does not require to `explicitly define types` of the Objects that you create.  

Basically what that means is that, `identifiers` and `Objects` are **separated**.
Consider this example:

```python
a = "String 1" # name 'a' refers to string object - str
a = 1 # now the same name 'a' refers to a different object - int.

a = [] # now same name 'a' refers to another different object - list.

# Now the only assingment that is left is  a = [].

# What do I want to say by that? 
# I want to say that what we call as 'variable' is actually an identifier/name that just refers to particular object.

# We can make one identifier refer to another identifier as well.
b = a # Now name 'b' refers to whatever object that name 'a' refers to.

# We can now see that 'a' and 'b' are refering to the same Object
print(a is b)
```

```python
# And we can also make 'b' refer to another object now:
b = 2

# And now a and b are not referring to the same Object.
print(a is b)
```

Objects can exist on their own, look:

```python
1 # 1 - is already a VALID Object, it has 'id' - "0x10000", it has 'type' - "int", it has 'value' - "1".
print(id(1))
print(type(1))

# Another example:

[1, 2, 3]
print(id([1, 2, 3]))
print(type([1, 2, 3]))
```

If you have looked at the **two** examples above, you might have noticed that
I repeated `1` **three** times, and `list of [1, 2, 3]` three times.

It naturally leads us to `Variables` -> `identifiers/names` that would `allow` us to `acceess` that structure and `pass` it around.

```python
# But remember, from the previous lesson, "identifiers/names" CAN NOT EXIST ON their own
p # See, name 'p' is not referring to any Object. That is why Python raises NameError Exception.
```

```python
p = 10 # They must refer to "some" Object. And to make them refer to an Object, we use "Assignment" operator: = .
```

I hope that, you at least understood something from this, to keep it short:  
The thing you should remember is that: 
**identifiers** `MUST HAVE` some object that they `refer` to, while objects can exist on their own.

# Type

Now, let's switch to the `second` component of the Object - **type**.

What is a `type`? What is `type` of an Object?

This is an **abstract concept**, not very clear what is it. However in **practice**, in `reality`, it is much **simpler**.
Why is it simple? Because `EVERY ONE OF US`, every single human, even a kid knows what `type` is, just doesn't realize that.

Put it simply: `type` of an Object is `what can we do` with this object, what `operations`, `actions` can we do with this Object.

Now many things become clear: why do you `sit` on a `chair`, why do you not `lie` on it?   
Why do you use a `pen` to `write` something down on a paper and not use a `chair` to `write` that down?

The examples might be **silly**, but they showcase the fundamental aspect of our `MIND`. The concept of `type` is **naturally** engraved in our `MIND` it is formed from an early stage of our development.

The whole purpose of these examples above was to actually reason about what `operations` can we do with this particular Object of type 'str'.

Just think for a second, we could just take a `pen` and `paper`, and we could write **text** on it.   
`Whatever` came into our mind. And we could write that text `however` we `wanted`.  
We could write some parts all uppercase, most of it as lowercase, but remember the grammar rules?   
Each sentence should start with `Capital Letter`!

Developers who are designing programming languages for us to use, are taking those concepts from `Reality` and then **figuring out** the way to make the `machine` be able to imitate the real world patterns.

```python
# Now think about Numbers as well:

num = 5 # This is a number, we studied basic math, so we know that we can add, subtract, multiply and divide the numbers.

# Does the language allow us to do so here? Because in real world, we can absolutely do them all.

num2 = num + 1

num2 # see this, we can actually "add" numbers as well.
```

```python
list_of_num1 = [1] # let's create some list - container

list_of_num2 = [2] # another one

# Let's try to add two lists!

print(list_of_num1 + list_of_num2) # See this, we were able to add things to the list as well!

# What about strings? Can we add them as well?

string1 = "String1"
string2 = "String2"

print(string1 + string2)
```

Okay we have seen now what a `correct type` means, but what is the `wrong type`?   
Engrave in your mind, that whenever Python sees the `wrong type` it raises `TypeError` Exception!

Let's consider this:

```python
# We certainly can not add text with number, it just does not make sense.

string1 = "String1"

num1 = 1

string1 + num1 # See, Python raises TypeError, why? 
# We have already seen with our own eyes, that both number and strings supported addition!!!
# The reason for this error, is that we tried to add two "unrelated objects" with "different" and "incompatible" types!
```

```python
# We certainly can not subtract one text from another!
string1 = "String1"
string2 = "String2"

string1 - string2
```

```python
# We also can not use "division" concept on a text! It does not make sense!
string1 // string2
```

```python
# However it is perfectly fine to subtract one number from another and divide one number over the other!
num1 = 10
num2 = 2

print(num1 - num2) # 8
print(num1 // num2) # 5
```

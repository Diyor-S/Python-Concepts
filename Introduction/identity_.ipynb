Today, we would be talking about Object's `identity`.
We would be talking about `Immutable` and `Mutable` Objects.

First things first, we have already scratched `id` of the Object during previous lessons,  
now we need to understand this better.

Object's `identity` defines `uniqueness` of that Object. Why is this `uniqueness` so important?
That is because it defines Object's `state`.

Consider this example:

```python
ten_years = 10

eleven_years = 11

# Look, both of these Object's type is the same.
type(ten_years) == type(eleven_years)
```

```python
# However if we check whether these objects are the same or not, we get False
ten_years is eleven_years
```

Another example:

This is a good example of Identifiers and Objects are separated!

```python
num10 = 10 # Identifier 'num10' references the Object of type 'int' and value '10'

print(id(num10)) # Let's see its identity

num10 + 1 # Let's now try to update its value!

print(id(num10)) # The object stayed the same, as it should.
print(num10) # However, look at what happened, the addition did not take any effect!
```

```python
# Objects of type 'int' are considered to be 'Immutable' objects.

num10 = 10 # We initialized an identifier num10 to reference Object with value 10!
# Python already allocated memory for the Object of type 'int' and value '10'

print(id(num10)) # We see the address of the Object to which name 'num10' points

num10 = num10 + 1 # Careful here! We are actually updating where name 'num10' points to!!!
# We have overriden the Object to which our identifier 'num10' points to!

print(id(num10)) # We will see a different Address!
```

Output above gave different `Addresses`. In other words, this means that Object's `identity` has changed after performing the `math operation`.

Now, what is actually `Immutable` Object?

An `Immutable` Object is an Object such that when tried to **perform** `operation` on it, it `no longer exists`!

Meaning if you ever want to change that Object, it would mean that it is gonna `die` and a memory space would be allocated for it in replacement.

```python
number = 1000

# Let's try to perform operation on it
number -= 100

print(number) # We see that our Object's value actually changed!
# But again, Careful here!
```

```python
# Repeating the same experiment with checking `identity` before and after operation!
number = 1000

print(id(number), number)


number -= 100

print(id(number), number)

# If you look careful at the output, the Object's `identity` has changed! They are different Objects now!
```

```python
# Now if we try to save the result
number = 1000


result = number - 100

# And check whether they are the same Objects, it would be False!
number is result
```

```python
# The same is applied to `str` Objects. 
# They are also `Immutable`, so each change brings `new` object to be created and prior gets destroyed (garbage collected).

string1 = "Hello"

# Let's try to perform `operation` on 'str' Object as well.
string2 = string1.lower()

print(string1, string2) # Different objects.
print(string1 is string2) # So it outputs false
```

```python
# We can even check it like this:

# Look here we see 4 different identities, tho the values produced are the same.

# But in fact those on right side which just printed the value, they are also different objects. 
# So there are 8 different Objects, not 4, and all have the SAME 'type' and the 'value'

print(id(string1.lower()), string1.lower()) 
print(id(string1.lower()), string1.lower())
print(id(string1.lower()), string1.lower())
print(id(string1.lower()), string1.lower())
```

To sum up, we have covered the Immutable Objects, common `data types` which are considered `immutable`: int, str, float, boolean, tuple.

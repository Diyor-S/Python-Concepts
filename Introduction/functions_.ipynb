What is a `function`?  

First of all it is an `Object`.   
It has all the same things like: **identity**, **type**, and the **value**.  

Second of all, it is **conceptually** different from easily grasped Objects like `int`, `str`.  

Before we continue let's see what the function looks like:

````python
# This is the basic form of the 'function'. 
# We can already see that it has 'special syntax'.
def func1():
    return 1

func1() 
````

In the above example, there are two **stages**: `Function **Definition**`, and `Function **Invocation**`.

You may ask how did you know what **keyword** is used to define a function?   
Besides the overall internet videos, articles, the most reliable way to discover this information is via **official documentation**.   
Because there are things that we can find out there.

https://docs.python.org/3/reference/compound_stmts.html#function-definitions


```python
funcdef:                   [decorators] "def" funcname [type_params] "(" [parameter_list] ")"
                           ["->" expression] ":" suite
```

This explain what keyword can be used, there are parentheses ( and ) that we can use that are `REQUIRED`.  
"" - `REQUIRED`  
[] - `OPTIONAL`  
paramter_list - wrapped with [] so it is optional, meaning function can have 0 args with () usage is mandatory.  
suite - function body where we write what the function would do when it gets **invoked**.  

So now, let's see the function's: identity, type, and value in practice:

````python
# Function Definition 
def func2(): 
    # Function Body
    return "This is function 2"

func2 # Function is an Object so its "value" is: <function __main__.func2()> - the reference to the function Object itself.
````

````python
id(func2) # Function Object's identity: 128737892376768
````

````python
type(func2) # Function Object's type: function
````

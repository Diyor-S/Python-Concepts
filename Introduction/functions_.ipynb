### Functions, series 2/?.

We have explored the first stage of the functions: `Function Definition`.   
It includes: using a `keyword - def`, giving a `name (identifier)` to the `function Object`,   
and using `parentheses` with optionally `paramter list`, and using `:` semi-colon to end `function's **signature**`.  

So basically everything until `:` semi-colon is considered to be `Function Signature`.

Then, follows the `Function's **Body**`.   
This where we write down what the `function object` will **execute** when it gets `invoked`.  

Function definition would not be `any useful` if there was `no second stage`: `Function Invocation` stage.

What is `Invocation`?  
It means **calling** the **function Object**: using operator `()` on the function **name (identifier)**.

`Function Body` - is only **executed** when the function gets **invoked**.

Let's consider a full example:

````python
# Function Definition Stage: 
# using 'def', giving a function 'name (identifier) = func1', using '()' and ':' to end 'Function Signature'
def func1():
    # 4-space identation
    # Function Body:
    print("This is the function 1")

    return 100 # Function invocation result.""
````

Now having finished Function Definition, it is up to developer - US,  
when, and where, and what we want to do with this function.  

````python
# Function Invocation:
func1() # using given 'name' together with the () - call operator.

# Look at the output, we have got Function's Body executed and it returned 100 as well.
````

So why is there `()` - operator to invoke the function?     
Because without the parentheses we use function as plain Object,   
meaning we use it just like any other Object that we have seen such as: numbers, strings, lists etc.  

What conclusion can we make from this?

Functions have two natures: **Object** nature, and **Behavioral** one.

Whenever we use the `name` itself without `()` operator, we are working with function's **Object** nature,   
on the contrary, when we use `name` together with `()` operator, we are dealing with function's **Behavioral** nature.
